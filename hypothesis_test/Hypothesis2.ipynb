```python
import numpy as np
import pandas as pd
from scipy.stats import wilcoxon

# ============================================================
# 1. Hybrid per-seed accuracies (30 seeds)
#    (from your "Hybrid Per Seed Results" table)
# ============================================================
hybrid_accuracy = np.array([
    0.9626, 0.9629, 0.9628, 0.9625, 0.9620, 0.9632, 0.9626, 0.9625, 0.9629, 0.9626,
    0.9625, 0.9628, 0.9632, 0.9624, 0.9628, 0.9624, 0.9626, 0.9625, 0.9631, 0.9627,
    0.9625, 0.9629, 0.9626, 0.9626, 0.9629, 0.9619, 0.9619, 0.9626, 0.9632, 0.9629
])

# ============================================================
# 2. Baseline models' per-seed accuracies (30 each)
#    (from your previous CNN / TCN / FNN / XGB / RF tables)
# ============================================================

# CNN (30-seed mean accuracies)
cnn_acc_scores = np.array([
    0.9357, 0.9366, 0.9362, 0.9360, 0.9363, 0.9363,
    0.9354, 0.9370, 0.9368, 0.9354, 0.9368, 0.9365,
    0.9354, 0.9358, 0.9364, 0.9361, 0.9357, 0.9367,
    0.9362, 0.9356, 0.9364, 0.9366, 0.9358, 0.9376,
    0.9346, 0.9368, 0.9359, 0.9368, 0.9371, 0.9361
])

# TCN
tcn_acc_scores = np.array([
    0.9315, 0.9322, 0.9322, 0.9331, 0.9333, 0.9319,
    0.9317, 0.9325, 0.9328, 0.9333, 0.9314, 0.9339,
    0.9333, 0.9327, 0.9327, 0.9325, 0.9313, 0.9325,
    0.9319, 0.9321, 0.9323, 0.9321, 0.9318, 0.9307,
    0.9304, 0.9317, 0.9315, 0.9322, 0.9333, 0.9307
])

# FNN
fnn_acc_scores = np.array([
    0.9359, 0.9354, 0.9369, 0.9358, 0.9356, 0.9356,
    0.9354, 0.9357, 0.9360, 0.9355, 0.9365, 0.9366,
    0.9364, 0.9360, 0.9356, 0.9362, 0.9364, 0.9372,
    0.9349, 0.9355, 0.9357, 0.9356, 0.9360, 0.9344,
    0.9354, 0.9359, 0.9360, 0.9352, 0.9364, 0.9349
])

# XGBoost
xgb_acc_scores = np.array([
    0.9539, 0.9532, 0.9541, 0.9533, 0.9535, 0.9536,
    0.9547, 0.9535, 0.9535, 0.9540, 0.9542, 0.9532,
    0.9537, 0.9543, 0.9533, 0.9533, 0.9538, 0.9536,
    0.9540, 0.9536, 0.9538, 0.9527, 0.9541, 0.9533,
    0.9539, 0.9534, 0.9535, 0.9541, 0.9538, 0.9532
])

# Random Forest
rf_acc_scores = np.array([
    0.9546, 0.9544, 0.9550, 0.9551, 0.9552, 0.9549,
    0.9547, 0.9542, 0.9545, 0.9547, 0.9549, 0.9544,
    0.9551, 0.9546, 0.9540, 0.9541, 0.9542, 0.9547,
    0.9547, 0.9547, 0.9544, 0.9547, 0.9545, 0.9546,
    0.9547, 0.9548, 0.9548, 0.9548, 0.9549, 0.9545
])

# ============================================================
# 3. Wilcoxon Signed-Rank Test Helper
#    H0: Hybrid accuracy is NOT significantly better than model
#    H1: Hybrid accuracy IS significantly better than model
# ============================================================
def compare_with_wilcoxon(hybrid_scores, baseline_scores, model_name, alpha=0.05):
    # Check same length
    if len(hybrid_scores) != len(baseline_scores):
        raise ValueError(f"Length mismatch: hybrid={len(hybrid_scores)}, {model_name}={len(baseline_scores)}")

    # Wilcoxon signed-rank test (paired, one-sided: hybrid > baseline)
    stat, p_value = wilcoxon(hybrid_scores, baseline_scores, alternative="greater")

    hybrid_mean = hybrid_scores.mean()
    base_mean   = baseline_scores.mean()
    mean_diff   = hybrid_mean - base_mean

    decision = "Reject H₀ (Hybrid significantly better)" if p_value < alpha else "Fail to reject H₀ (No significant improvement)"

    return {
        "Model Compared": model_name,
        "Hybrid Mean Accuracy": hybrid_mean,
        "Baseline Mean Accuracy": base_mean,
        "Mean Difference": mean_diff,
        "Wilcoxon Statistic": stat,
        "p-value": p_value,
        "Decision (α=0.05)": decision
    }

# ============================================================
# 4. Run Hypothesis 2 for all models
# ============================================================
results = []
results.append(compare_with_wilcoxon(hybrid_accuracy, rf_acc_scores,  "Random Forest"))
results.append(compare_with_wilcoxon(hybrid_accuracy, xgb_acc_scores, "XGBoost"))
results.append(compare_with_wilcoxon(hybrid_accuracy, cnn_acc_scores, "CNN"))
results.append(compare_with_wilcoxon(hybrid_accuracy, tcn_acc_scores, "TCN"))

results_df = pd.DataFrame(results)

# Pretty formatting
pd.set_option("display.precision", 6)
print("========== HYPOTHESIS 2 — HYBRID vs STANDALONE MODELS ==========")
results_df
```
